```python
#25.	Convert the hours logged column to a number.
df1["hours_logged"] = pd.to_numeric(df1["hours_logged"], errors="coerce")
```

```python
df1
```

```python
#Convert the billing rate column to a number. Remove the $ sign first.
df1["billing_rate_(usd)"] = (
    df1["billing_rate_(usd)"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)

df1["billing_rate_(usd)"] = pd.to_numeric(
    df1["billing_rate_(usd)"], errors="coerce"
)
```

```python
#27.	Convert the CSAT score column to a number.
df1["csat_score"] = pd.to_numeric(df1["csat_score"], errors="coerce")
```

```python


#28.	Print the average, the smallest and the largest value of hours logged.
print("Average:", df1["hours_logged"].mean())
print("Minimum:", df1["hours_logged"].min())
print("Maximum:", df1["hours_logged"].max())

```

```python
#29.	Sort the table by hours logged from highest to lowest and show the top 10 rows.
print(
    df1.sort_values("hours_logged", ascending=False).head(10)
)
```

```python
df1
```

```python
#31.	Remove the extra spaces from the client name column.
df1["client_name"] = df1["client_name"].str.strip()

print(df1["client_name"].head())
```

```python
#32.	Change the client name column to title case.
df1["client_name"] = df1["client_name"].str.title()

print(df1["client_name"].head())
```

```python

#33.	Convert the ticket created column to a date.
df1["ticket_created_at"] = pd.to_datetime(
    df1["ticket_created_at"],
    errors="coerce"
)

print(df1["ticket_created_at"].dtype)
```

```python
#30.	Show the 5 tickets with the highest billable amount.


#34.	Add a new column that holds the year of the ticket.
df1["ticket_year"] = df1["ticket_created_at"].dt.year

print(df1[["ticket_created_at", "ticket_year"]].head())

```

```python
#35.	Add a new column that holds the month of the ticket.
df1["ticket_month"] = df1["ticket_created_at"].dt.month

print(df1[["ticket_created_at", "ticket_month"]].head())

```

```python
#36.	Count how many tickets were created in each year.
print(df1["ticket_year"].value_counts().sort_index())

```

```python
#37.	Add a new column called total equal to billing rate multiplied by hours logged.
df1["total"] = df1["billing_rate_(usd)"] * df1["hours_logged"]

print(df1[["billing_rate_(usd)", "hours_logged", "total"]].head())
```

```python
#38.	Fill the missing values in the CSAT score column with the average CSAT score.
csat_mean = df1["csat_score"].mean()

df1["csat_score"] = df1["csat_score"].fillna(csat_mean)

print("Average CSAT used:", csat_mean)
print("Missing CSAT:", df1["csat_score"].isna().sum())

```

```python
#39.	Drop the rows where the client name is missing.
df1 = df1.dropna(subset=["client_name"])

print("Rows remaining:", len(df1))
```

```python
#40.	Group by service line and find the average hours logged.
print(
    df1.groupby("serviceline")["hours_logged"].mean().round(2)
)

```

```python
#41.	Group by priority and count the tickets.
df1.groupby('priority')['ticket_id'].count()
```

```python
#42.	Group by client name and find the total billable amount.

df1.groupby("client_name")["billable_amount_usd"].sum().sort_values(ascending=False)
```

```python
#43.	Find the top 5 clients by number of tickets.

print(
    df1["client_name"]
    .value_counts()
    .head(5)
)
```

```python
#44.	Find the average CSAT score for each priority.

print(
    df1.groupby("priority")["csat_score"].mean()
)
```

```python
#45.	Draw a bar chart of ticket count by service line.

import matplotlib.pyplot as plt

df1["serviceline"].value_counts().plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Ticket Count by Service Line")
plt.xlabel("Service Line")
plt.ylabel("Ticket Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#46.	Draw a histogram of hours logged.

plt.figure(figsize=(10, 6))

plt.hist(
    df1["hours_logged"].dropna(),
    bins=20,
    edgecolor="black"
)

plt.title("Distribution of Hours Logged")
plt.xlabel("Hours Logged")
plt.ylabel("Frequency")
plt.show()
```

```python
#47.	Show only the tickets where hours logged is more than 50.

print(df1[df1["hours_logged"] > 50])
```

```python
#48.	Show the tickets where status is Closed and priority is P1.

print(
    df1[
        (df1["status"] == "Closed") &
        (df1["priority"] == "P1")
    ]
)
```

```python
#49.	In the SLA breached column, change Yes, Y, TRUE and 1 to True, and everything else to False.

df1["sla_breached"] = (
    df1["sla_breached"]
    .astype(str)
    .str.strip()
    .str.upper()
    .isin(["YES", "Y", "TRUE", "1"])
)

print(df1["sla_breached"].value_counts())
```
